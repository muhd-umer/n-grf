```python
import sys

sys.path.append("../")

import argparse
import logging
from pathlib import Path

import torch
from torch.utils.tensorboard import SummaryWriter

from datasets.dataloader import get_wireless_dataloader
from models.gaussian_model import GaussianModel, GaussianModelConfig
from utils.general_utils import set_random_seed
from utils.train_utils import setup_logging
```

```python
class Config:
    data_path = "../datasets/outputs/conf_16x2_414u_5.0ghz_sbrRT_sc104.mat"
    batch_size = 16
    num_points = 16378
    sh_degree = 3
    use_pred_normals = False
    log_dir = "../logs"
    num_iterations = 200000
    checkpoint_freq = 5000
    seed = 42
    device = "cuda"
    tensorboard = False
    visualize = False
    vis_freq = 1000
    stl_path = "../datasets/models/conference.stl"
```

```python
def setup_experiment(config):
    """Setup experiment directory and logging"""
    log_dir = Path(config.log_dir)
    log_dir.mkdir(parents=True, exist_ok=True)

    logger = setup_logging(log_dir)

    writer = None
    if config.tensorboard:
        logger.info("Initializing TensorBoard writer...")
        writer = SummaryWriter(log_dir / "tensorboard")

    return logger, writer
```

```python
config = Config()
logger, writer = setup_experiment(config)
```

```python
logger.info("Initializing dataloader...")
dataloader = get_wireless_dataloader(
    config.data_path,
    batch_size=config.batch_size,
    num_pc=config.num_points,
    drop_last=True,
)

logger.info("Initializing model...")
model_config = GaussianModelConfig(
    sh_degree=config.sh_degree,
    max_sh_degree=config.sh_degree,
    use_pred_normals=config.use_pred_normals,
)
model = GaussianModel(model_config).to(config.device)
```

```python
# get initial point cloud and initialize model
batch = next(iter(dataloader))
point_cloud = batch["point_cloud"].to(config.device)
model.init_from_pc(point_cloud)

logger.info(f"Initialized model with {len(point_cloud)} Gaussians")

# training loop
logger.info("Starting training...")
for iteration in range(config.num_iterations):
    # ...existing training iteration code...

    if iteration % 100 == 0:
        if writer is not None:
            # ...log metrics using writer...
            pass
        # ...log to console/file...

    if iteration % config.checkpoint_freq == 0:
        # ...save checkpoint...
        pass
```
